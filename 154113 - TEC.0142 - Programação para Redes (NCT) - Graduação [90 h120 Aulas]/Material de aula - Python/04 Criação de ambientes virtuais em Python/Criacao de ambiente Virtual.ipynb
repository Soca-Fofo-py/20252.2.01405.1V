Ambientes virtuais criam "uma cópia" do Python e permitem  instalação de pacotes a ela restritos

###Criar um ambiente virtual


```
C:> cd BASE
C:> python -m venv nome_ambiente
```

###Ativar o ambiente virtual###

**Windows:**

```
c:> cd BASE
c:> nome_ambiente\Scripts\activate.bat
```

**Linux:**

```
$ cd BASE
$ source nome_ambiente/bin/activate`
```

###Desativar o ambiente virtual###

**Windows:**

```
c:> cd BASE
c:> nome_ambiente\Scripts\deactivate.bat
```

**Linux:**

```
$ exit
```
